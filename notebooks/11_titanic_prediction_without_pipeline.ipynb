### Introduction to Pipelines
---

* **Pipelines** chain together multiple steps so that the output of each step is used as input to the next step.
* **Pipelines** make it easy to apply the same preprocessing to train and test!


```python
import pandas as pd
import numpy as np
import pickle
```

```python
ohe_sex = pickle.load(open('models/ohe_sex.pkl', 'rb'))
ohe_embarked = pickle.load(open('models/ohe_embarked.pkl', 'rb'))
clf = pickle.load(open('models/clf.pkl', 'rb'))
```

```python
# Assume user input 
# Pclass/gender/Age/SiSb/Parch/Fare/Embarked
test_input = np.array([1, 'male', 45.5, 0, 0, 28.5, 'S'], dtype=object).reshape(1,7)
```

```python
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
```

```python
test_input_sex
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:,-1].reshape(1,1))
```

```python
test_input_embarked
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_transformed = np.concatenate((test_input[:,[0,3,4,5]], test_input_age, test_input_sex, test_input_embarked), axis=1)
```

```python
test_input_transformed.shape
```

```python
clf.predict(test_input_transformed)
```
